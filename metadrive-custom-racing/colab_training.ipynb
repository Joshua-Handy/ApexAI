### What is VecNormalize?

**VecNormalize normalizes observations during training** - this is critical for RL!

- **Without it**: Observations have different scales (speed: 0-120, position: -100 to 100, etc.)
- **With it**: All observations normalized to similar ranges (mean=0, std=1)
- **Result**: Agent learns 3-4x faster and more reliably

**The notebook automatically:**
- ✅ Enables VecNormalize during training (`--vecnorm`)
- ✅ Saves normalization stats to `.pkl` files
- ✅ Loads stats when resuming training
- ✅ Includes stats in download package

**Important:** Always keep your `vecnorm_*.pkl` files with your models!

## Monitoring Tips

**During training, you'll see:**
```
| time/total_timesteps  | 50000     |
| rollout/ep_len_mean   | 1523.2    | ← Episode length (higher = better)
| rollout/ep_rew_mean   | 145.3     | ← Average reward (higher = better)
```

**Good signs:**
- `ep_len_mean` increasing (agent surviving longer)
- `ep_rew_mean` increasing (agent getting better rewards)
- Episodes reaching 2000+ steps = completing laps

**Bad signs:**
- `ep_len_mean` stuck at 200-500 (crashing early)
- `ep_rew_mean` stuck negative (not learning)

**"No GPU available"**

- Restart runtime

**"Cannot find module 'metadrive'"**
- Re-run Step 1 (install dependencies)

**"File not found: train_multi_custom_speedway.py"**
- Re-upload your code folder
- Check Step 3 - make sure you're in /content/metadrive-custom-racing/examples

**Session disconnected during training**
- Find latest checkpoint: `!ls -lah ../agents/`
- Resume from checkpoint using resume_training.py

**Training very slow**
- Free T4 is slower - this is normal
- Consider upgrading to Colab Pro for A100
- Or use Kaggle/RunPod instead
