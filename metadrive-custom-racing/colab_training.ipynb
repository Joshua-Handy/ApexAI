## Monitoring Tips

**During training, you'll see:**
```
| time/total_timesteps  | 50000     |
| rollout/ep_len_mean   | 1523.2    | ← Episode length (higher = better)
| rollout/ep_rew_mean   | 145.3     | ← Average reward (higher = better)
```

**Good signs:**
- `ep_len_mean` increasing (agent surviving longer)
- `ep_rew_mean` increasing (agent getting better rewards)
- Episodes reaching 2000+ steps = completing laps

**Bad signs:**
- `ep_len_mean` stuck at 200-500 (crashing early)
- `ep_rew_mean` stuck negative (not learning)

- Re-upload your code folder
